```python
def rgb2gray(name):
    I=tm.ImageAsArray(name,'preprocessDB')
    # If image is already float (0-1), scale it up to 255 first
    if I.dtype == np.float32 or I.dtype == np.float64:
         I = (I * 255).astype(np.uint8)
    return I
```
