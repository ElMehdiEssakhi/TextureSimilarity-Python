# Optimization: Precomputing Feature Vectors

## 1. Objective
To improve search efficiency, this workflow moves the computationally expensive step of feature extraction out of the query loop. By calculating feature vectors for the entire dataset beforehand, subsequent search queries become significantly faster.

```python
import thisModule as tm
import distances as ds
```

```python
features = tm.preComputeAllImagesFeaturesVector(tm.getImages(66))
```

Query image ID 46 against the precomputed feature set

```python
result=ds.distance_texture_precomputed(46, features)
```
